We are building a small machine-learning application that identifies patterns in simulated motor vibration.

Imagine a motor running under three conditions:
- Normal: the motor vibrates slightly.
- Imbalance-like: the vibration amplitude is higher.
- Excessive vibration: the signal contains stronger vibration and additional frequency components.
We generate these signals using Python, calculate three features, and train a machine-learning model to classify them.
